```python
import pandas as pd


advanced_stats = pd.read_csv("C:/Users/Zagros/Desktop/mini3/advanced_stats.csv")


advanced_stats["team"] = advanced_stats["team"].fillna("TOT")


advanced_stats = advanced_stats.sort_values(
    ["season", "player_id", "team"],
    key=lambda s: s != "TOT"
)


advanced_stats = advanced_stats.drop_duplicates(["season", "player_id"], keep="first")

advanced_stats.to_csv("C:/Users/Zagros/Desktop/mini3/data clean/advanced_stats.csv", index=False)



```
