```python
from sqlalchemy import Column, Integer, String, Float, create_engine 
from sqlalchemy.orm import declarative_base, sessionmaker
from sqlalchemy.ext.declarative import declarative_base
Base = declarative_base()
```

```python
#MVP
Session = sessionmaker(bind=engine)
session = Session()
class MVP_c(Base):
    __tablename__ = 'MVP'

    id = Column(Integer, primary_key=True, autoincrement=True)
    Season = Column(String(255))
    League = Column(String(255))
    Player_id = Column(String(255))
    Age = Column(Integer)
    Team_id = Column(String(255))
    Game = Column(Integer)
    Minutes_Played_Per_Game = Column(Float)
    Points_Per_Game = Column(Float)
    Total_Rebounds_Per_Game = Column(Float)
    Assists_Per_Game = Column(Float)
    Steals_Per_Game = Column(Float)
    Blocks_Per_Game = Column(Float)
    Field_Goal_Percentage = Column(Float)
    Three_Point_Field_Goal_Percentage = Column(Float)
    Free_Throw_Percentage = Column(Float)
    Win_Shares = Column(Float)
    Win_Shares_Per_48_Minutes = Column(Float)


Base.metadata.create_all(engine)
```

```python
MVP = [
    MVP_c(
        Season=row['Season'],
        League=row['League'],
        Player_id=row['Player_id'],
        Age=int(row['Age']) if pd.notna(row['Age']) else None,
        Team_id=row['Team_id'],
        Game=int(row['Game']) if pd.notna(row['Game']) else None,
        Minutes_Played_Per_Game=float(row['Minutes Played Per Game']) if pd.notna(row['Minutes Played Per Game']) else None,
        Points_Per_Game=float(row['Points Per Game']) if pd.notna(row['Points Per Game']) else None,
        Total_Rebounds_Per_Game=float(row['Total Rebounds Per Game']) if pd.notna(row['Total Rebounds Per Game']) else None,
        Assists_Per_Game=float(row['Assists Per Game']) if pd.notna(row['Assists Per Game']) else None,
        Steals_Per_Game=float(row['Steals Per Game']) if pd.notna(row['Steals Per Game']) else None,
        Blocks_Per_Game=float(row['Blocks Per Game']) if pd.notna(row['Blocks Per Game']) else None,
        Field_Goal_Percentage=float(row['Field Goal Percentage']) if pd.notna(row['Field Goal Percentage']) else None,
        Three_Point_Field_Goal_Percentage=float(row['3-Point Field Goal Percentage']) if pd.notna(row['3-Point Field Goal Percentage']) else None,
        Free_Throw_Percentage=float(row['Free Throw Percentage']) if pd.notna(row['Free Throw Percentage']) else None,
        Win_Shares=float(row['Win Shares']) if pd.notna(row['Win Shares']) else None,
        Win_Shares_Per_48_Minutes=float(row['Win Shares Per 48 Minutes']) if pd.notna(row['Win Shares Per 48 Minutes']) else None
    )
    
    for _, row in dmvp.iterrows()
]
session.add_all(MVP)
session.commit()
session.close()
```
